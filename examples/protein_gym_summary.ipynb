```python
import pandas as pd
from scipy.stats import spearmanr

import glob
import os
```

```python
SCORE_DIR = "/hpc/group/naderilab/eleanor/AMPLIFY_ALLY/ProteinGym/output"
```

```python
score_csv_files = sorted(glob.glob(os.path.join(SCORE_DIR, "*.csv")))

scc_ls = []
dms_fname = []
for file in score_csv_files:
    df = pd.read_csv(file)
    rho, pval = spearmanr(df["RITA_score"], df["DMS_score"])
    scc_ls.append(rho)
    dms_fname.append(file.split('/')[-1])

scc_df = pd.DataFrame({"File": dms_fname, "SCC": rho})
```
