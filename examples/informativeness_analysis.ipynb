## Anlysis on the relationship between lambda and sequence

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
DIR = "/hpc/group/naderilab/eleanor/AMPLIFY_ALLY/logs/unconstrained_seed.100/summary/"
```

```python
with open(os.path.join(DIR, "lambdas.json")) as f:
    data = json.load(f)

df = pd.DataFrame.from_dict(data, orient="index", columns=["sequence", "lambda"])
df.index.name = "record_id"
df = df.reset_index()
```

```python
df["seq_len"] = df["sequence"].str.len()

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df["seq_len"], df["lambda"], alpha=0.3, s=5, color="steelblue")
ax.set_xlabel("Sequence Length")
ax.set_ylabel("Lambda")
ax.set_title("Lambda vs Sequence Length")
plt.tight_layout()
# plt.savefig("lambda_vs_seqlen.png", dpi=150)
plt.show()
```
